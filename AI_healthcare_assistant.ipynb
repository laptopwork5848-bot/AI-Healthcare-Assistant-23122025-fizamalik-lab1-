```python
import getpass
api_key =getpass.getpass("enter your gemini api key:")
```

```python
from google import genai
client=genai.Client(api_key=api_key)
print("gemini connect successfully")
```

```python
question = input("Enter your health question: ")

prompt = f"""
You are an AI Healthcare Assistant.
Understand the user's question and give simple general health information.
Do not give a medical diagnosis.
If the situation seems serious, advise the user to contact a healthcare professional.

User's question:
{question}
"""

print("Question received successfully!")
```

```python
response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=prompt
)

print("\nAI Healthcare Assistant:")
print(response.text)
```
